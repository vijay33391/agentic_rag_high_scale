```python
from dotenv import load_dotenv 
import os
load_dotenv()
```

```python
from langchain_groq import ChatGroq
from nemoguardrails import RailsConfig, LLMRails

# Our primary LLM for all experiments
groq_llm = ChatGroq(
    api_key=os.environ.get("GROQ_API_KEY"),
    model="llama-3.3-70b-versatile",
    temperature=0
) 

print("Groq LLM ready: llama-3.3-70b-versatile")
```

```python
# just a simple example of greeings guadils
user_greeting="""
def user express greeting():
    "hi"
    "hello"
define bot express greeting
  "Hello! I'm your Enterprise IT Assistant. I specialise in Kubernetes, Intel hardware, and enterprise networking. What can I help you with today?"

define flow greeting
  user express greeting
  bot express greeting
"""

YAML_BASE = """
models:
  - type: main
    engine: openai
    model: gpt-3.5-turbo

instructions:
  - type: general
    content: |
      You are an Enterprise IT Assistant specialising in:
      - Kubernetes (deployment, scaling, operators, networking)
      - Intel hardware (CPUs, FPGAs, NICs, SRIOV)
      - Enterprise networking (SDN, VLANs, BGP, routing)
      Only answer questions about these topics. Be professional and concise.
"""
```

```python
config = RailsConfig.from_content(colang_content=user_greeting, yaml_content=YAML_BASE)
```

```python
rails = LLMRails(config=config, llm=groq_llm)
user_input = "hello"
```

```python
import nest_asyncio

nest_asyncio.apply()
```

```python
response = rails.generate(messages=[{"role": "user", "content": user_input}])
print(f"User: {user_input}")
print(f"Bot: {response}")
```
